````python
%matplotlib inline
%load_ext autoreload
%autoreload 2
import matplotlib.pyplot as plt
import xarray
````

# Load dataset

````python
print('Load dataset')
nc1 = xarray.load_dataset('./test_semic_transect_AlbDenby_893a942.nc')
nc2 = xarray.load_dataset('./test_semic_transect_AlbDenby_4da5301.nc')
````

# Plot difference between SEMIC-fortran and SEMIC-cpp

````python
fig, axs = plt.subplots(2,2,figsize=(10,6))

for nrow, stnid in enumerate([0, 4]):
    print('Plot (a) SMB')
    ax = axs[nrow,0]
    nc1.isel(nx=stnid).smb.plot(ax=ax,label='SEMIC-fortran',color='b')
    nc2.isel(nx=stnid).smb.plot(ax=ax,label='SEMIC-cpp',color='r')
    ax.legend()
    ax.set_ylabel('SMB (m ice yr-1)')
    ax.set_xlabel('time (days)')

    print('Plot (b) SMB')
    ax = axs[nrow,1]
    (nc2.isel(nx=stnid).smb-nc1.isel(nx=stnid)).smb.plot(ax=ax,color='k')
    ax.set_ylabel('$\Delta$SMB (m ice yr-1)')
    ax.set_xlabel('time (days)')
    fig.tight_layout()
````
